# Calculate Payoffs

```python
#true performance in 11 bins
performance = pd.DataFrame(columns=['score', 'share'])

for i in range(11):
    share = 100*sum(1 for x in delegator['participant.overall_score'] if x==i)/len(delegator['participant.overall_score'])
    performance.loc[i] = [i, share]
```

```python
performance
```

## evaluator payoff

```python
evaluator['task_payoff'] = np.nan
evaluator['rand_belief'] = np.nan
evaluator['belief_difficulty_payoff'] = 0

for row in range(evaluator.shape[0]):
    #How much did the delegator earn them?
    if evaluator['match_outcome'][row] == 1:
        #good outcome
        evaluator.loc[row, 'task_payoff'] = evaluator['good_outcome_eva'][row]
    else:
        #bad outcome
        evaluator.loc[row, 'task_payoff'] = evaluator['bad_outcome_eva'][row]
    
    #How much did the evaluator earn in the belief elicitation?
    evaluator.loc[row, 'rand_belief'] = random.randint(0, 10)
    true_share_eva = performance[performance['score']==evaluator['rand_belief'][row]]['share']
    if (abs(int(true_share_eva) - evaluator['difficulty' + str(int(evaluator['rand_belief'][row]))][row]) <= 5):
        evaluator.loc[row, 'belief_difficulty_payoff'] = evaluator['belief_payment_difficulty'][row]


evaluator['final_payoff'] = evaluator['task_payoff'] + evaluator['belief_difficulty_payoff'] + evaluator['risk_task_payoff']
evaluator['bonus_payoff'] = evaluator['final_payoff'] - 1
```

```python
evaluator['bonus_payoff'].mean()
```

## delegator payoff

```python
delegator['task_payoff'] = np.nan
delegator['confidence_payoff'] = np.nan
delegator['rand_belief'] = np.nan
delegator['belief_difficulty_payoff'] = 0

for row in range(delegator.shape[0]):
    #Task payoff
    if ast.literal_eval(delegator['participant.relevant_round'][row])['success']:
        delegator.loc[row, 'task_payoff'] = delegator['good_outcome_del'][row]
    else:
        delegator.loc[row, 'task_payoff'] = delegator['bad_outcome_del'][row]
    
    #Belief about own performance payoff
    delegator.loc[row, 'confidence_payoff'] = delegator['pay_belief'][row]*delegator['belief_payment_confidence'][row]
    
    #Belief about difficulty
    delegator.loc[row, 'rand_belief'] = random.randint(0, 10)
    true_share = performance[performance['score']==delegator['rand_belief'][row]]['share']
    if (abs(int(true_share) - delegator['difficulty' + str(int(delegator['rand_belief'][row]))][row]) <= 5):
        delegator.loc[row, 'belief_difficulty_payoff'] = delegator['belief_payment_difficulty'][row]
            
delegator['bonus_payoff'] = delegator['task_payoff'] + delegator['confidence_payoff'] + delegator['belief_difficulty_payoff'] + delegator['risk_task_payoff']
```
