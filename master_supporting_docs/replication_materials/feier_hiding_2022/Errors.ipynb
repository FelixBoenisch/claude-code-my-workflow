```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# load CSV data
```

```python
names = ["Subject", "Treatment", "Errors"]
```

```python
data = pd.read_csv("Fehler_Task.csv", names = names,  sep = ",")
```

```python
# check CSV data
```

```python
data.head
```

```python
# check mean and std of human treatment
```

```python
human = data.loc[data['Treatment'] == 1]
```

```python
human["Errors"].mean()
```

```python
human["Errors"].std()
```

```python
# check mean and std of machine treatment
```

```python
machine = data.loc[data['Treatment'] == 2]
```

```python
machine["Errors"].mean()
```

```python
machine["Errors"].std()
```

```python
# check mean and std of all subjects
```

```python
data["Errors"].mean()
```

```python
data["Errors"].std()
```

```python
# check if means between treatments differ significantly
```

```python
stat, p = stats.ttest_ind(human["Errors"], machine["Errors"])
print('Statistics=%.3f, p=%.3f' % (stat, p))
```
