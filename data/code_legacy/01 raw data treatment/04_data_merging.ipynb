# Load Data

### Delegator

```python
import pandas as pd
import numpy as np

dg_pilot = pd.read_excel('../../interim/01_cleaned/delegator_pilot.xlsx')
dg_main = pd.read_excel('../../interim/01_cleaned/delegator_main.xlsx')
dg_no_p = pd.read_excel('../../interim/01_cleaned/delegator_main_no_punish.xlsx')
dg_no_p2 = pd.read_excel('../../interim/01_cleaned/delegator_main_no_punish2.xlsx')
```

```python
#unify columns
s1 = set(dg_pilot.columns)
s2 = set(dg_main.columns)
s3 = set(dg_no_p.columns)
s4 = set(dg_no_p2.columns)

for i in list(s1-s3):
    dg_no_p[i] = np.nan

for i in list(s3-s1):
    dg_pilot[i] = np.nan
    
for i in list(s3-s2):
    dg_main[i] = np.nan
    
for i in list(s1-s2):
    dg_main[i] = np.nan

s3 = set(dg_no_p.columns)
s4 = set(dg_no_p2.columns)

for i in list(s3-s4):
    dg_no_p2[i] = np.nan

#unify column order
columns = dg_pilot.columns
dg_main = dg_main[columns]
dg_no_p = dg_no_p[columns]
dg_no_p2 = dg_no_p2[columns]

#add pilot and treatment variable
dg_pilot['pilot']=1
dg_main['pilot']=0
dg_no_p['pilot']=0
dg_no_p2['pilot']=0

dg_pilot['treat']=0
dg_main['treat']=0
dg_no_p['treat']=1
dg_no_p2['treat']=1
```

```python
dg = pd.concat([dg_pilot, dg_main, dg_no_p, dg_no_p2])
```
