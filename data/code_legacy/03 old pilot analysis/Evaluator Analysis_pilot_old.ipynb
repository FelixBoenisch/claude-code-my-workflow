```python
import pandas as pd
import statsmodels.api as sm
import numpy as np
from stargazer.stargazer import Stargazer

pd.options.display.max_rows = 4000
ev = pd.read_excel('../../interim/01_cleaned/evaluator_main.xlsx')
ev = ev[ev['attention']==0.5].reset_index()
```

#### Difficulty

```python
#calculate weighted average
ev['wa_difficulty'] = np.nan
for i in range(ev.shape[0]):
    wa = 0
    for j in range(11):
        wa += (ev['difficulty' + str(j)][i]/100)*j
    ev.loc[i, 'wa_difficulty'] = wa
```

```python
ev['wa_difficulty'].describe()
```

```python
ev['bad_good_del'] = ev['punish_del_bad'] - ev['punish_del_good']
ev['bad_good_nodel'] = ev['punish_nodel_bad'] - ev['punish_nodel_good']
ev['nodel_del_good'] = ev['punish_nodel_good'] - ev['punish_del_good']
ev['nodel_del_bad'] = ev['punish_nodel_bad'] - ev['punish_del_bad']
```

```python
ev[['bad_good_del', 'bad_good_nodel', 'nodel_del_good', 'nodel_del_bad']].describe()
```

```python
ev_3 = pd.DataFrame(np.repeat(ev.values, 3, axis=0), columns=ev.columns)

#ev delete never punish
ev_p = ev[[not i for i in ((ev['punish_del_good']==0) & (ev['punish_del_bad']==0) & (ev['punish_nodel_bad']==0) & (ev['punish_nodel_good']==0))]]
ev_p_3 = pd.DataFrame(np.repeat(ev_p.values, 3, axis=0), columns=ev_p.columns)
```

```python
import scipy
print(scipy.stats.mannwhitneyu(ev_3['punish_del_bad'].astype(float), ev_3['punish_nodel_bad'].astype(float)))
print(scipy.stats.mannwhitneyu(ev_p['punish_del_bad'].astype(float), ev_p['punish_nodel_bad'].astype(float)))
print(scipy.stats.mannwhitneyu(ev_p_3['punish_del_bad'].astype(float), ev_p_3['punish_nodel_bad'].astype(float)))

print(scipy.stats.mannwhitneyu(ev_3['punish_del_good'].astype(float), ev_3['punish_nodel_good'].astype(float)))
print(scipy.stats.mannwhitneyu(ev_p['punish_del_good'].astype(float), ev_p['punish_nodel_good'].astype(float)))
print(scipy.stats.mannwhitneyu(ev_p_3['punish_del_good'].astype(float), ev_p_3['punish_nodel_good'].astype(float)))

print(scipy.stats.ttest_ind(ev_3['punish_del_bad'].astype(float), ev_3['punish_nodel_bad'].astype(float)))
print(scipy.stats.ttest_ind(ev_p['punish_del_bad'].astype(float), ev_p['punish_nodel_bad'].astype(float)))
print(scipy.stats.ttest_ind(ev_p_3['punish_del_bad'].astype(float), ev_p_3['punish_nodel_bad'].astype(float)))

print(scipy.stats.ttest_ind(ev_3['punish_del_good'].astype(float), ev_3['punish_nodel_good'].astype(float)))
print(scipy.stats.ttest_ind(ev_p['punish_del_good'].astype(float), ev_p['punish_nodel_good'].astype(float)))
print(scipy.stats.ttest_ind(ev_p_3['punish_del_good'].astype(float), ev_p_3['punish_nodel_good'].astype(float)))
```
