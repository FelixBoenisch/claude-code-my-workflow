```python
import pandas as pd
pd.set_option('display.float_format', '{:.0f}'.format)
```

# Delegator

```python
dg = pd.read_excel('../../interim/02_appended/delegator_full.xlsx')
```

```python
pt_pre_pilot = pd.read_csv('../../raw/pre_pilot/delegator/PageTimes-2023-02-26.csv')
pt_pilot1 = pd.read_csv('../../raw/pilot/delegator/PageTimes-2023-02-26.csv')
pt_pilot2 = pd.read_csv('../../raw/pilot/delegator/PageTimes-2023-02-26-2.csv')
pt_main = pd.read_csv('../../raw/main/delegator/PageTimes-2023-06-19.csv')
pt_main_np = pd.read_csv('../../raw/main/delegator_no_punish/PageTimes-2023-06-19.csv')
pt_main_np2 = pd.read_csv('../../raw/main/delegator_no_punish/PageTimes-2023-06-20.csv')
pt_main_np3 = pd.read_csv('../../raw/main/delegator_no_punish_2/PageTimes-2023-06-21.csv')
```

```python
#create wide DFs
wide_dfs = []

for pt in [pt_pre_pilot, pt_pilot1, pt_pilot2, pt_main, pt_main_np, pt_main_np2, pt_main_np3]:
    #wide_format
    pt_wide = pt.pivot_table(index='participant_code', columns=['page_name', 'round_number'], values='epoch_time_completed')
    #concatenate
    wide_dfs.append(pt_wide)

wide = pd.concat(wide_dfs)
```

```python
zero = wide.columns.get_level_values(0)
one = wide.columns.get_level_values(1)

columns = []
for item1, item2 in zip(zero, one):
    if item1 == 'Round':
        columns.append(f"{item1}{item2}")
    else:
        columns.append(f"{item1}")
        
wide.columns = columns
```

```python
#merge with dg and remove duplicates
merged_df = pd.merge(dg, wide, left_on='participant.code', right_on='participant_code', how='left')
merged_df_unique = merged_df.drop_duplicates()
```

```python
#save
merged_df_unique.to_excel('../../interim/03_pagetimes/delegator_full_merged.xlsx', index=False)
```
